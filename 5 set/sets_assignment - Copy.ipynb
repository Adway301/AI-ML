```python
#1 creating and accessing set
s1=set(range(0,11))
print(s1)
```

```python
#2 creating and removing elements
s1=set(range(0,11))
print(s1)
s1.add(11)
s1.remove(0)
print(s1)
```

```python
#3 set operation
s=set(range(0,6))
s1=set(range(0,10,2))
print(s)
print(s1)
union=s.union(s1)
intersection=s.intersection(s1)
difference=s.difference(s1)
symmetric_difference=s.symmetric_difference(s1)
print(f"{union, intersection, difference, symmetric_difference}" )
```

```python
#4 set comprehension
square={i**2 for i in range(1,10)}
print(square)
```

```python
#5 filtering sets
s1=set(range(1,11))
s={i for i in s1 if i%2==0}
print(s)
```

```python
#6 set methods
s={1,4,7,2,4,2,5,6,44,4,1}
set=set(s)
print(set)
```

```python
#7 subset and superset 
s1={i for i in range(0,5)}
s2={i for i in range(0,3)}
subset=s2.issubset(s1)
print(subset)
superset=s1.issuperset(s2)
print(superset)
```

```python
#8 frozen set
f_s=frozenset(range(0,6))
print(f_s)
```

```python
#9 set and list conversion
s=set(range(0,6))
print(s)
l=list(s)
print(l)
l.append(6)
print(set(l))
```

```python
#10 set and dictionary
s={
    "s1":"set1",
    "s2":"set2",
    "s3":"set3"

}
print((s))
```

```python
#11 iterating over sets
s=set(range(1,6))
print(s)
for i in s:
    print(i)
```

```python
#12 removing elements from set
s=set(range(1,5))
print(s)
while s:
    s.pop()
    print(s)
s.discard(5)
print(s)
```

```python
#13 set symmetric difference update
s1={2,4,6,2,5,6,3,4,5,6,4,3,3,3}
s2={2,4,6,8,9,3,4,5,7,8,4}
s1=s1.symmetric_difference(s2)
print(s1)
```

```python
#14 set membership testing
s={1,3,5,6,8,3,6,3,7,4,7,4,5}
print(3 in s)
print(43 in s)
```
