```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from urllib.parse import urljoin
```

```python
URL="https://quotes.toscrape.com/"
```

```python
# if res.status_code==200:
#     print(res.status_code)
# with open("scrapped_data/quotes.html", "w",encoding="utf-8") as f:
#         f.write(res.text)
```

```python

quotes_by_author=[]
all_quote=soup.find_all("div",class_="quote")


while URL:
    res=requests.get(URL)
    soup=BeautifulSoup(res.text,"html.parser")

    #Extract quote and author
    for quote in all_quote:
        quotes=quote.find("span",class_="text").get_text(strip=True)
        authors=quote.find("small", class_="author").get_text(strip=True)

        # Extrach all tags
        tags=quote.find_all("a",class_="tag")
        tag_list=[]
        for tag in tags:
            tag_list.append(tag.get_text(strip=True))

        # Check if life exists in list
        if "life" in tag_list:
            clean_tags=",".join(tag_list)
            quotes_by_author.append([quotes,authors,clean_tags])

    next_btn=soup.select_one("li.next a") 

    if next_btn:
        URL=urljoin(URL,next_btn["href"])
    else:
        URL=None    
df=pd.DataFrame(quotes_by_author,columns=["quote", "author","tags"])
df
```

```python
df.to_csv("cleaned_data/clean_prac.csv",index=False)
```
