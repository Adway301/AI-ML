```python
#1 create and access tuple
tp=tuple(range(0,5))
print(tp)
```

```python
#2 print first middle and last element in tuple
print(tp[0])
print(tp[len(tp)//2])
print(tp[-1])
```

```python
#3 tuple slicing

print(tp[:3])
print(tp[-3:])
print(tp[2:5])
```

```python
#4 nested tuple
matrix=(
    (1,4,2),
    (6,3,9,),
    (2,9,5)
)

for row in matrix:
    print(row)
print({matrix[1][2]})
```

```python
#5 tuple concatenation
tp1=tuple(range(0,3))
tp2=tuple(range(3,7))
t3=tp1+tp2
print(t3)
```

```python
#6 duplicates in tuple
tp=(1,3,4,5,31,1,4,5,14,5,4,5)
print(tp)
tp.count(5)
tp.index(3)
```

```python
#7 unpacking tuples
tp=(1,2,3,4,5)
a,b,c,d,e=tp
print(a,b,c,d,e)
```
