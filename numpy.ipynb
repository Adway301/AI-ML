```python
import numpy as np
import time
```

```python
size=100000
arr_list = list(range(size))
start = time.time()
for i in arr_list:
    i**2
end = time.time()
print(f"time in normal list is {end - start} seconds")


arr2= np.array(arr_list)
start = time.time()
arr2=arr2**2
end = time.time()
print(f"time for numpy calculation is {end-start} seconds")
```
