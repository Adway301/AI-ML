```python
#3 if elif else
a=int(input("Enter a:"))
if a>0:
    print("positive")
elif a<0:
    print("negative")
else:
    print("zero")
```

```python
#12 factorial calculation
a=int(input("enter number:"))
factorial = 1
i=1

while i<a:
    factorial=factorial*i
    i=i+1
print(f"The factorial of {a} is {factorial}.")
```

```python
#14 check prime number
num=int(input("Enter number:"))
is_prime=True
if num <=1:
    is_prime=False
else:
    for i in range(2, num):
        if num%i==0:
            is_prime=False
            
if is_prime:
    print(f"{num} is prime number.")
else:
    print(f"{num} is not prime number.")          


```

```python
#15 fibonacci sequence
i=int(input("Enter a:"))
a,b=0,1
count=0
while count<i:
    print(a)
    a,b=b,a+b
    count+=1
    


```
