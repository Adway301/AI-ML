```python
#1 create and access dictionary
d={i:i**2 for i in range(1,11)}
print(d)
```

```python
#2 accessing dictionary elements
print(d[5])
print(d.keys())
```

```python
#3 dictionary methods
d[11]=121
d.pop(1)
print(d)
```

```python
#4 iterating over dictionaries
for key,value in d.items():
    print(f"{key}:{value}")
```

```python
#5 dictionary comprehension
cubes={i:i**3 for i in range(1,11)}
print(cubes)
```

```python
#6 merging dictionaries
square1={i:i**2 for i in range(1,6)}
square2={i:i**2 for i in range(6,11)}
square1.update(square2)
print(square1)
```

```python
#7 nested dictionaries
student={
    "name":"adway",
     "age":21,
     "grade":{"maths":21,"science":21,"english":22}
     }
print(student)

```

```python
#8 dictionary of list
d={ i:[i*j for j in range(1,6)] for i in range(1,6)}
print(d)
```

```python
#9 dictionary of tuple
d={i:(i,i**2) for i in range(1,6)}
print(d)
```

```python
#10 dictionary and list conversion
d={i:i**2 for i in range(1,6)}
print(list(d.items()))
```

```python
#11 dictionary filtering
d={i:i**2 for i in range(1,11)}
print(d)
d={i:i**2 for i in range(1,11) if i%2==0}
print(d)
```

```python
#12 dictionary key and value transformation
d={i:i**2 for i in range(1,6)}
print(d)
d1={v:k for k,v in d.items()}
print(d1)
```
